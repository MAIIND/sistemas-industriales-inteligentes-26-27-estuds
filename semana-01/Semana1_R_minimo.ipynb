## 1. Variables y tipos básicos

Una **variable** es un nombre que guarda un valor. En R se asigna con `<-` (se lee "guarda esto en"). También funciona `=`, pero lo habitual en R es `<-`.

````r
silice <- 2.35           # un número decimal (numeric): % de sílice
numero_columna <- 3L     # un número entero (integer): la L lo marca como entero
etapa <- "flotacion"     # un texto (character), entre comillas
fuera_spec <- TRUE       # un lógico: TRUE o FALSE (en mayúsculas)

print(silice)
print(etapa)
print(fuera_spec)
````

Cada valor tiene un **tipo**. Puedes consultarlo con `class()`:

````r
print(class(silice))         # numeric   -> número decimal
print(class(numero_columna)) # integer   -> número entero
print(class(etapa))          # character -> texto
print(class(fuera_spec))     # logical   -> lógico (booleano)
````

Diferencia con Python: en R un número escrito sin más, como `3`, es **decimal** (`numeric`). Solo es entero si le pones la `L` (`3L`). En la práctica casi nunca importa.

Una variable puede cambiar de valor: la última asignación es la que vale.

````r
x <- 10
x <- x + 5    # coge el valor actual de x (10), le suma 5, y lo vuelve a guardar en x
print(x)      # 15
````

## 2. Estructuras de datos

Dos formas de guardar varios valores juntos: el **vector** (el equivalente de la lista de Python) y la **lista** (el equivalente del diccionario).

### Vectores `c( )`
Un **vector** es una colección ordenada de valores **del mismo tipo**. Se crea con `c()` (de *combinar*).

````r
silices <- c(2.35, 1.80, 3.10, 2.05)

print(silices)
print(length(silices))   # length() da cuántos elementos tiene: 4
````

Accedes a un elemento por su **posición**. **Atención, diferencia importante con Python: en R se empieza a contar en 1, no en 0.**

````r
print(silices[1])                  # el primero: 2.35
print(silices[2])                  # el segundo: 1.80
print(silices[length(silices)])    # el último: 2.05
````

**Trampa para quien viene de Python:** en R un índice negativo **no** cuenta desde el final, sino que **quita** ese elemento. `silices[-1]` no es el último: es todo el vector **menos el primero**.

````r
print(silices[-1])   # todos menos el primero: 1.80 3.10 2.05
````

Para añadir un elemento al final, se combina el vector con el valor nuevo y se guarda el resultado:

````r
silices <- c(silices, 3.90)   # añade al final
print(silices)
````

### Listas con nombres `list( )`
Una **lista con nombres** guarda parejas **nombre → valor**, y los valores pueden ser de tipos distintos. Es el equivalente del diccionario de Python. Accedes por el nombre con `$` (o con `[["nombre"]]`).

````r
registro <- list(hora = "01:00", silice_media = 2.35, fuera_spec = TRUE)

print(registro$hora)              # accedo por el nombre: "01:00"
print(registro[["silice_media"]]) # lo mismo con doble corchete: 2.35
````

También se puede empezar con una lista **vacía** e ir añadiendo elementos uno a uno:

````r
fila <- list()                  # lista vacía
fila$hora <- "02:00"            # añado el elemento 'hora'
fila$silice_media <- 1.80       # añado otro
fila$silice_max <- 3.40

str(fila)                       # str() muestra la estructura de forma compacta
````

## 3. Operaciones y comparaciones

Operaciones aritméticas normales:

````r
print(10 + 3)     # 13
print(10 - 3)     # 7
print(10 * 3)     # 30
print(10 / 3)     # 3.333...
````

**Comparaciones:** preguntan algo y devuelven un lógico (`TRUE` o `FALSE`). Comparar es `==` (doble igual), como en Python.

````r
print(5 == 5)     # ¿son iguales? TRUE
print(5 == 8)     # FALSE
print(5 < 8)      # ¿menor que? TRUE
print(5 > 8)      # FALSE
print(5 != 8)     # ¿distintos? TRUE
````

**Lo más característico de R: todo funciona sobre vectores enteros.** Una operación o una comparación sobre un vector se aplica a **cada elemento a la vez**, sin escribir ningún bucle. En Python esto solo pasa con pandas o numpy; en R pasa siempre:

````r
silices <- c(2.35, 1.80, 3.10, 2.05)

print(silices * 10)    # multiplica cada elemento: 23.5 18.0 31.0 20.5
print(silices > 3)     # compara cada elemento: FALSE FALSE TRUE FALSE
````

Ese vector de `TRUE`/`FALSE` es la base de los **filtros** que usaremos con las tablas ("dame las filas donde la sílice sea mayor que 3").

## 4. Bucles `for`

Un bucle `for` repite algo para cada elemento de una colección. Lo que se repite va entre llaves `{ }`:

````r
silices <- c(2.35, 1.80, 3.10)

for (s in silices) {
  cat("sílice:", s, "\n")
}
````

En cada vuelta, la variable `s` toma el valor del siguiente elemento. El nombre `s` lo eliges tú.

`cat()` escribe varios valores seguidos en la misma línea (`"\n"` es el salto de línea). `print()` muestra un solo objeto.

**Recorrer dos vectores a la vez.** Como R no tiene desempaquetado, se recorre la **posición** `i` y se usa para leer los dos vectores. `seq_along(horas)` da las posiciones 1, 2, 3…:

````r
horas  <- c(1, 2, 3)
medias <- c(2.35, 1.80, 3.10)

for (i in seq_along(horas)) {
  cat("la hora", horas[i], "tuvo una sílice media de", medias[i], "\n")
}
````

En R los bucles se usan **mucho menos** que en Python: como las operaciones ya funcionan sobre vectores enteros (sección 3), casi siempre hay una forma sin bucle. Aun así conviene entenderlos.

## 5. Condicionales `if`

Un `if` ejecuta algo solo si una condición es cierta. La condición va entre paréntesis y el código entre llaves; `else if` y `else` cubren los demás casos.

````r
silice <- 3.2

if (silice > 3.0) {
  print("fuera de especificación")
} else if (silice > 2.0) {
  print("en el límite")
} else {
  print("dentro de especificación")
}
````

**Ojo con dónde va el `else`:** escríbelo siempre en la misma línea que la llave que cierra (`} else {`). Si lo pones en la línea siguiente, en una celda como esta R da error.

**Elegir un valor u otro en una línea.** En R el propio `if` devuelve un valor, así que se puede usar a la derecha de una asignación:

````r
silice <- 3.2

fuera_spec <- if (silice > 3.0) 1 else 0
print(fuera_spec)   # 1, porque la condición es cierta
````

Ese `if` solo sirve para **un** valor. Para convertir una condición en 1 o 0 sobre un **vector** entero (lo que necesitarás en la Semana 1 para la columna `fuera_spec`) hay dos formas vectorizadas:

- `ifelse(condición, valor_si_cierto, valor_si_falso)`, que decide elemento a elemento;
- `as.integer(condición)`, que convierte `TRUE`/`FALSE` en 1/0 (el equivalente del `.astype(int)` de pandas).

````r
silices <- c(2.35, 3.40, 1.80, 3.10)

print(ifelse(silices > 3.0, 1, 0))   # 0 1 0 1
print(as.integer(silices > 3.0))     # 0 1 0 1, lo mismo
````

## 6. Funciones: usarlas y construirlas

Una **función** es una operación con nombre a la que le das unos datos (**argumentos**, entre paréntesis) y te devuelve un resultado. Ya has usado algunas:

````r
numeros <- c(3, 8, 1, 6)

print(length(numeros))   # cuántos elementos hay -> 4
print(sum(numeros))      # la suma -> 18
print(max(numeros))      # el mayor -> 8
print(min(numeros))      # el menor -> 1
print(mean(numeros))     # la media -> 4.5 (en Python base no existe; en R sí)
````

Diferencia con Python: en R **no hay métodos con punto** como `nombre.upper()` o `columna.mean()`. Todo son funciones a las que se les pasa el valor: `toupper(nombre)`, `mean(columna)`. En R el punto es una letra más dentro de los nombres (por eso hay funciones como `read.csv` o `as.integer`).

````r
nombre <- "lunes"
print(toupper(nombre))   # convierte a mayúsculas -> "LUNES"
````

### Construir tus propias funciones

Cuando una operación se repite, o quieres darle un nombre que diga qué hace, la defines tú con `function`, y la guardas en una variable como cualquier otro valor:

````r
# Devuelve 1 si la sílice supera el límite de especificación (3,0 %), 0 si no.
fuera_de_spec <- function(silice) {
  if (silice > 3.0) {
    return(1)
  } else {
    return(0)
  }
}

print(fuera_de_spec(3.2))   # 1
print(fuera_de_spec(2.4))   # 0
````

Las piezas:

- el **nombre**, `<-`, la palabra `function` y los **parámetros** entre paréntesis;
- el **cuerpo**, entre llaves `{ }` (en R la sangría es solo para leer mejor; lo que manda son las llaves);
- `return(...)` **devuelve** el resultado, con paréntesis, y la función termina ahí;
- el comentario de encima explica qué hace la función. No es obligatorio, pero es una buena costumbre.

Definir la función no ejecuta nada: solo se ejecuta cuando la **llamas**, como `fuera_de_spec(3.2)`.

**El `return` se puede omitir.** Si no lo escribes, la función devuelve **el valor de la última línea** que ejecuta. Es muy habitual en R:

````r
media <- function(valores) {
  sum(valores) / length(valores)    # última línea: es lo que se devuelve
}

m <- media(c(2.4, 3.1, 2.2))
print(m)   # 2.5666...
````

**Ojo: mostrar no es devolver.** `cat()` solo muestra un valor en pantalla y devuelve `NULL` (nada). Si es lo último que hace la función, la función devuelve `NULL`:

````r
media_que_muestra <- function(valores) {
  cat(sum(valores) / length(valores), "\n")   # solo la muestra
}

m1 <- media_que_muestra(c(2.4, 3.1, 2.2))   # muestra 2.566667, pero no la devuelve
m2 <- media(c(2.4, 3.1, 2.2))               # no muestra nada, pero m2 guarda el valor

print(m1)   # NULL
print(m2)   # 2.566667
````

**Parámetros con valor por defecto.** Si un parámetro casi siempre vale lo mismo, puedes darle un valor por defecto. Quien llama a la función solo lo indica si quiere otro valor:

````r
# 1 si la sílice supera el límite, 0 si no. Límite por defecto: 3,0 %.
fuera_de_spec <- function(silice, limite = 3.0) {
  if (silice > limite) 1 else 0
}

print(fuera_de_spec(3.2))               # usa el límite por defecto -> 1
print(fuera_de_spec(3.2, limite = 3.5)) # con otro límite -> 0
````

**Devolver varios valores.** Una función de R devuelve **un solo** objeto. Para devolver varios resultados se devuelve una **lista con nombres** (sección 2), y se saca cada uno por su nombre:

````r
# Devuelve el mínimo, la media y el máximo de un vector de números.
resumen <- function(valores) {
  list(minimo = min(valores), media = mean(valores), maximo = max(valores))
}

r <- resumen(c(2.4, 3.1, 2.2, 3.3, 2.5))
cat(r$minimo, r$media, r$maximo, "\n")
````

**Usar tu función como cualquier otra.** Una vez definida, se llama igual que `length` o `max`; por ejemplo, dentro de un bucle `for`:

````r
silices <- c(2.35, 3.40, 1.80, 3.10)

for (s in silices) {
  cat(s, "->", fuera_de_spec(s), "\n")
}
````

**Trampa: un `if` dentro de una función no funciona con un vector entero.** `fuera_de_spec(silices)` da error, porque `if` solo acepta **una** condición y le estarías pasando cuatro. Hay dos salidas: aplicar la función a cada elemento con `sapply()`, o escribirla de forma vectorizada con `ifelse()` (sección 5), que es más corto y más rápido:

````r
print(sapply(silices, fuera_de_spec))   # aplica la función a cada elemento: 0 1 0 1

fuera_de_spec_vec <- function(silice, limite = 3.0) {
  ifelse(silice > limite, 1, 0)         # vectorizada: funciona con un valor o con un vector
}
print(fuera_de_spec_vec(silices))       # 0 1 0 1
````

**Lo de dentro se queda dentro.** Los parámetros y las variables que creas dentro de una función solo existen mientras se ejecuta; desde fuera no se ven. Por eso todo lo que la función necesita debe **entrar por los parámetros** y todo lo que produce debe **salir por lo que devuelve**.

### `dplyr`: el equivalente de pandas

Con R base se puede hacer todo, pero para agrupar y resumir tablas lo habitual es la librería **`dplyr`**. Sus funciones son verbos que se leen casi como una frase. Se carga con `library()` (el equivalente del `import` de Python). Los avisos de que algunas funciones quedan "enmascaradas" (*masked*) son normales: `dplyr` tiene su propia `filter`, por ejemplo, y pasa a ser la que se usa.

````r
library(dplyr)
````

**La tubería `|>`.** Pasa lo que tiene a la izquierda como primer argumento de la función de la derecha. Se lee "y después". Estas dos líneas hacen lo mismo:

````r
head(tabla, 2)
tabla |> head(2)    # "coge tabla y después quédate con las 2 primeras filas"
````

Con `dplyr`, filtrar y crear columnas se escribe así (dentro de los verbos de `dplyr` las columnas se nombran sin `tabla$`):

````r
tabla |> filter(silice > 3)                              # filtrar filas
tabla |> mutate(fuera_spec = as.integer(silice > 3))     # crear una columna
````

**Agrupar: `group_by()` + `summarise()`.** Es la herramienta central de la Semana 1. `group_by()` junta las filas que comparten el mismo valor (aquí, la misma `hora`) y `summarise()` calcula un resumen de cada grupo. El resultado tiene **una fila por hora**. Cada resumen lleva el nombre que tú le das:

````r
tabla |>
  group_by(hora) |>                       # 1. agrupa las filas que tienen la misma hora
  summarise(silice_media = mean(silice))  # 2. una fila por hora, con la media de la sílice
````

Fíjate en que la `hora` sale como **una columna normal**. Aquí no hay nada parecido al índice de pandas: el resultado de `dplyr` es una tabla como cualquier otra (equivale al `groupby(..., as_index=False)` de pandas). El resultado de `summarise()` es un *tibble*, la versión de tabla de `dplyr`: se usa igual que un data frame y se muestra con el tipo de cada columna debajo del nombre (`<chr>` texto, `<dbl>` número decimal, `<int>` entero).

**Varios resúmenes a la vez.** Dentro de `summarise()` pones tantos como quieras, separados por comas:

- `mean()`, `max()`, `min()`: los de siempre;
- `first()`: el **primer** valor de cada grupo (útil cuando una columna es constante dentro del grupo);
- `n()`: cuántas **filas** tiene cada grupo.

Así se monta de una vez una tabla resumen con solo las columnas que te interesan. Es el patrón de la sección 4 de la Semana 1:

````r
resumen <- tabla |>
  group_by(hora) |>
  summarise(
    silice_media = mean(silice),
    silice_max   = max(silice),
    columna_1    = first(columna),
    n_filas      = n()
  )

resumen
````

**La media de todas las columnas a la vez.** Cuando hay muchas columnas (en la Semana 1, más de 20), no se escriben una a una: `across()` aplica la misma función a varias columnas. `where(is.numeric)` elige todas las columnas numéricas:

````r
tabla |>
  group_by(hora) |>
  summarise(across(where(is.numeric), mean))
````

**Qué hace `group_by` por dentro.** Para entender el mecanismo, el R base tiene `split()`, que reparte las filas de la tabla en una lista de mini-tablas, una por cada valor de la columna. Recorriendo esa lista con un bucle haces a mano lo que `group_by` + `summarise` hacen solos:

````r
grupos <- split(tabla, tabla$hora)   # lista con nombres: un data frame por cada hora

for (valor_hora in names(grupos)) {
  grupo <- grupos[[valor_hora]]
  cat("hora", valor_hora, "\n")
  print(grupo)
  cat("sílice media de esa hora:", mean(grupo$silice), "\n")
  cat("-----\n")
}
````

El bucle da tres vueltas (14:00, 17:00 y 19:00); en cada una, `grupo` es la mini-tabla con solo las filas de esa hora, y `mean(grupo$silice)` es la media de **esa** hora. En la Semana 1 no escribirás este bucle: usarás `group_by` + `summarise`, que hace lo mismo para todas las columnas a la vez.
