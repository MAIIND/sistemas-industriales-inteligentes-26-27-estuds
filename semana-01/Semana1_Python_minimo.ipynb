## 1. Variables y tipos básicos

Una **variable** es un nombre que guarda un valor. Se asigna con `=` (que aquí significa "guarda esto en", no "es igual a").

```python
silice = 2.35            # un número decimal (float): % de sílice
numero_columna = 3      # un número entero (int)
etapa = "flotacion"     # un texto (string), entre comillas
fuera_spec = True       # un booleano: True o False

print(silice)
print(etapa)
print(fuera_spec)
```

Cada valor tiene un **tipo**. Puedes consultarlo con `type()`:

```python
print(type(silice))         # float -> número decimal
print(type(numero_columna)) # int   -> número entero
print(type(etapa))          # str   -> texto
print(type(fuera_spec))     # bool  -> booleano
```

Una variable puede cambiar de valor: la última asignación es la que vale.

```python
x = 10
x = x + 5     # coge el valor actual de x (10), le suma 5, y lo vuelve a guardar en x
print(x)      # 15
```

## 2. Estructuras de datos

Tres formas de guardar varios valores juntos. Las tres aparecen en el curso.

### Listas `[ ]`
Una **lista** es una colección ordenada de valores. Se escribe entre corchetes.

```python
silices = [2.35, 1.80, 3.10, 2.05]

print(silices)
print(len(silices))     # len() da cuántos elementos tiene: 4
```

Accedes a un elemento por su **posición** (empezando en 0):

```python
print(silices[0])    # el primero: 2.35
print(silices[1])    # el segundo: 1.80
print(silices[-1])   # el último: 2.05
```

Puedes añadir elementos con `.append()`:

```python
silices.append(3.90)   # añade al final
print(silices)
```

### Diccionarios `{ }`
Un **diccionario** guarda parejas **clave → valor**. En vez de acceder por posición, accedes por la clave. Se escribe entre llaves.

```python
registro = {'hora': '01:00', 'silice_media': 2.35, 'fuera_spec': True}

print(registro['hora'])          # accedo por la clave: '01:00'
print(registro['silice_media'])  # 2.35
```

Muchas veces se empieza con un diccionario **vacío** y se van añadiendo claves una a una — útil para ir construyendo un registro a mano, fila a fila:

```python
fila = {}                       # diccionario vacío
fila['hora'] = '02:00'          # añado la clave 'hora'
fila['silice_media'] = 1.80     # añado otra clave
fila['silice_max'] = 3.40

print(fila)
```

### Tuplas `( )`
Una **tupla** es como una lista pero que no se modifica. Aparecen sobre todo al "desempaquetar" varios valores a la vez:

```python
punto = (3, 2.35)     # una tupla con dos valores

# desempaquetar: repartir los dos valores en dos variables de golpe
columna, valor = punto
print(columna)   # 3
print(valor)     # 2.35
```

Ese desempaquetado (recibir dos valores separados por coma) es la clave para entender el bucle del `groupby`, que verás en la sección 4 y 7.

## 3. Operaciones y comparaciones

Operaciones aritméticas normales:

```python
print(10 + 3)     # 13
print(10 - 3)     # 7
print(10 * 3)     # 30
print(10 / 3)     # 3.333...
```

**Comparaciones:** preguntan algo y devuelven un booleano (`True` o `False`). Ojo: comparar es `==` (doble igual), porque el `=` simple ya se usa para asignar.

```python
print(5 == 5)     # ¿son iguales? True
print(5 == 8)     # False
print(5 < 8)      # ¿menor que? True
print(5 > 8)      # False
print(5 != 8)     # ¿distintos? True
```

## 4. Bucles `for`

Un bucle `for` repite algo para cada elemento de una colección. Recorrer una lista:

```python
silices = [2.35, 1.80, 3.10]

for s in silices:
    print("sílice:", s)
```

En cada vuelta, la variable `s` toma el valor del siguiente elemento. El nombre `s` lo eliges tú.

**Bucle con dos variables (desempaquetado).** Si cada elemento es una pareja, puedes recibir sus dos partes a la vez. Esto es exactamente lo que hace el bucle del `groupby`:

```python
parejas = [(1, 2.35), (2, 1.80), (3, 3.10)]

for hora, valor in parejas:
    print("la hora", hora, "tuvo una sílice media de", valor)
```

Aquí, en cada vuelta, `hora` recibe el primer valor de la pareja y `valor` el segundo. Este mismo desempaquetado es el que usa `groupby` por dentro (lo verás en la sección 7): cada grupo se entrega como una pareja (la etiqueta del grupo, y sus filas).

## 5. Condicionales `if`

Un `if` ejecuta algo solo si una condición es cierta. `elif` (else-if) y `else` cubren los demás casos.

```python
silice = 3.2

if silice > 3.0:
    print("fuera de especificación")
elif silice > 2.0:
    print("en el límite")
else:
    print("dentro de especificación")
```

**Expresión corta `if ... else`.** A veces quieres asignar un valor u otro según una condición, en una sola línea. Se lee "esto **si** se cumple, **si no** aquello":

```python
alta_silice = True

# vale 1 si la sílice es alta (fuera de spec), 0 si no
fuera_spec = 1 if alta_silice else 0
print(fuera_spec)
```

Este mismo patrón —convertir una condición en 1 o 0— es el que necesitarás en la Semana 1 para la característica `fuera_spec`. Allí, en vez de escribir el `if` fila a fila, pandas lo hace de una vez sobre toda la columna con `.astype(int)` sobre una condición booleana (lo verás en la sección 7). Aquí, primero, la versión escalar —fila a fila, con un solo valor—:

```python
silice = 3.2

fuera_spec = 1 if silice > 3.0 else 0
print(fuera_spec)   # 1, porque la condición es cierta
```

## 6. Funciones: usarlas y construirlas

### Construir tus propias funciones

Cuando una operación se repite, o quieres darle un nombre que diga qué hace, la defines tú con `def`. En el curso lo harás desde la Semana 2 (por ejemplo, el descenso de gradiente programado a mano) y lo verás en casi todos los cuadernos. Esta es la estructura:

```python
def fuera_de_spec(silice):
    """Devuelve 1 si la sílice supera el límite de especificación (3,0 %), 0 si no."""
    if silice > 3.0:
        return 1
    else:
        return 0

print(fuera_de_spec(3.2))   # 1
print(fuera_de_spec(2.4))   # 0
```

Las piezas:

- `def`, el **nombre** de la función, los **parámetros** entre paréntesis y dos puntos.
- El **cuerpo**, indentado (4 espacios), igual que en un `for` o un `if`.
- `return` **devuelve** el resultado a quien llamó a la función, y la función termina ahí.
- El texto entre triples comillas (*docstring*) explica qué hace la función. No es obligatorio, pero es una buena costumbre.

Definir la función no ejecuta nada: solo se ejecuta cuando la **llamas**, como `fuera_de_spec(3.2)`.

**Ojo: `print` no es `return`.** `print` solo muestra un valor en pantalla; `return` lo devuelve para que lo puedas guardar en una variable y seguir usándolo. Una función sin `return` devuelve `None` (nada):

```python
def media_que_imprime(valores):
    print(sum(valores) / len(valores))     # solo la muestra

def media(valores):
    return sum(valores) / len(valores)     # la devuelve

m1 = media_que_imprime([2.4, 3.1, 2.2])    # muestra 2.566..., pero no la devuelve
m2 = media([2.4, 3.1, 2.2])                # no muestra nada, pero m2 guarda el valor

print(m1)   # None
print(m2)   # 2.5666...
```

**Parámetros con valor por defecto.** Si un parámetro casi siempre vale lo mismo, puedes darle un valor por defecto. Quien llama a la función solo lo indica si quiere otro valor:

```python
def fuera_de_spec(silice, limite=3.0):
    """1 si la sílice supera el límite, 0 si no. Límite por defecto: 3,0 %."""
    return 1 if silice > limite else 0

print(fuera_de_spec(3.2))               # usa el límite por defecto -> 1
print(fuera_de_spec(3.2, limite=3.5))   # con otro límite -> 0
```

**Devolver varios valores.** Una función puede devolver varios resultados separados por comas. En realidad devuelve una tupla, que recoges con el desempaquetado de la sección 2:

```python
def resumen(valores):
    """Devuelve el mínimo, la media y el máximo de una lista de números."""
    return min(valores), sum(valores) / len(valores), max(valores)

minimo, media_silice, maximo = resumen([2.4, 3.1, 2.2, 3.3, 2.5])
print(minimo, media_silice, maximo)
```

**Usar tu función como cualquier otra.** Una vez definida, se llama igual que `len` o `max`; por ejemplo, dentro de un bucle `for`:

```python
silices = [2.35, 3.40, 1.80, 3.10]

for s in silices:
    print(s, "->", fuera_de_spec(s))
```

**Lo de dentro se queda dentro.** Los parámetros y las variables que creas dentro de una función solo existen mientras se ejecuta; desde fuera no se ven. Por eso todo lo que la función necesita debe **entrar por los parámetros** y todo lo que produce debe **salir por el `return`**.

## 7. pandas mínimo

`pandas` es la librería para trabajar con **tablas** de datos. Se importa así (el `as pd` es solo un apodo para escribir menos):

```python
import pandas as pd
```

Una **tabla** en pandas se llama *DataFrame*. Normalmente se carga de un fichero con `pd.read_csv('fichero.csv')`, pero aquí crearemos una pequeña a mano para practicar:

```python
tabla = pd.DataFrame({
    'hora':    ['14:00', '14:00', '17:00', '17:00', '19:00'],
    'columna': [1, 2, 1, 2, 1],
    'silice':  [2.4, 3.1, 2.2, 3.3, 2.5]
})

tabla
```

**Ver las primeras filas** con `.head()`. Sin nada entre paréntesis muestra las 5 primeras; con un número, las que le pidas. Con esta tabla de 5 filas no se nota, pero con las 20.000 del fichero de la Semana 1 es la forma de echar un vistazo sin llenar la pantalla:

```python
tabla.head(3)    # las 3 primeras filas
```

**Seleccionar una columna** con corchetes y su nombre:

```python
print(tabla['silice'])        # la columna silice
print()
print(tabla['silice'].mean()) # su media
```

**Filtrar filas** según una condición. Dentro de los corchetes va una comparación, y pandas se queda con las filas donde es cierta:

```python
# filas donde la sílice es mayor que 3
altas = tabla[tabla['silice'] > 3]
altas
```

**Crear una columna a partir de una condición.** Esta es la forma vectorizada que anunciaba la sección 5: pandas evalúa la comparación en toda la columna a la vez, y `.astype(int)` convierte `True`/`False` en 1/0. También puedes aplicar tu función de la sección 6 fila a fila con `.apply()`. El resultado es el mismo, pero la forma vectorizada es más corta y más rápida, y es la que usarás en la Semana 1:

```python
tabla['fuera_spec'] = (tabla['silice'] > 3).astype(int)            # vectorizada
tabla['fuera_spec_apply'] = tabla['silice'].apply(fuera_de_spec)    # con tu función

tabla
```

**`groupby`: agrupar por el valor de una columna.** Es la herramienta central de la Semana 1. Agrupa las filas que comparten el mismo valor (aquí, la misma `hora`) para resumir cada grupo.

Por ejemplo, la sílice media de cada hora, en tres pasos:

```python
grupos = tabla.groupby('hora')        # 1. agrupa las filas que tienen la misma hora
silice_grupos = grupos['silice']      # 2. de cada grupo, quédate con la columna silice
silice_grupos.mean()                  # 3. calcula la media de cada grupo

# Lo mismo en una sola instrucción, encadenando los tres pasos (así lo verás a menudo):
# tabla.groupby('hora')['silice'].mean()
```

**Cómo leer el resultado.** Parece que hay dos columnas, pero solo hay **una**: la de las medias (`2.75`, `2.75`, `2.50`). Lo que hay a la izquierda (`14:00`, `17:00`, `19:00`) son las horas, y aparecen como el **índice** del resultado.

**El índice es el nombre de cada fila**, como los números de fila a la izquierda de una hoja de Excel, pero con nombres que pueden significar algo (aquí, la hora). Encima de los nombres aparece de dónde salen (`hora`), y debajo, qué se ha calculado (`Name: silice`). Se lee así: *«a las 14:00 la sílice media fue 2,75; a las 17:00, 2,75; a las 19:00, 2,50»*.

**Los datos están en las columnas; el índice solo pone nombre a las filas.** Por eso no se comporta como una columna:

- no aparece en la lista de columnas, y `resultado['hora']` daría error;
- no cuenta en el tamaño (`shape`);
- no entra en los cálculos: `.mean()` no calcula «la media de la hora».

Para lo que sí sirve es para **buscar una fila por su nombre**, con `.loc[nombre]`. Compruébalo con todas las columnas a la vez:

```python
por_hora = tabla.groupby('hora').mean()
print(por_hora)
print()
print('Columnas:', list(por_hora.columns))   # 'hora' no está: no es una columna
print('Índice:  ', list(por_hora.index))     # las horas son los nombres de las filas
print('Tamaño:  ', por_hora.shape)           # (filas, columnas): el índice no cuenta
print()
print(por_hora.loc['17:00'])                 # buscar una fila por su nombre: la de las 17:00
```

**¿De dónde sale el nombre de cada fila?** Depende de cómo se haya creado la tabla:

| Cómo se crea la tabla | Nombre de las filas (índice) |
|---|---|
| Al crearla o cargarla de un fichero (`pd.DataFrame(...)`, `pd.read_csv(...)`) | Un entero por defecto: `0, 1, 2…`, la posición de cada fila. Es el que ves a la izquierda de `tabla`. |
| Al **filtrar** o seleccionar filas (`tabla[tabla['silice'] > 3]`) | Las filas **conservan el nombre que tenían** en la tabla original. Por eso `altas`, más arriba, tenía las filas `1` y `3`, no `0` y `1`. |
| Con **`groupby`** | El **valor del grupo**: una fila por cada valor distinto de la columna por la que agrupas (aquí, cada hora). |
| Con `groupby(..., as_index=False)` | Un entero por defecto, `0, 1, 2…`, y la columna de agrupación se queda como columna normal. |
| Con **`.reset_index()`** | Renumera las filas `0, 1, 2…`. El índice antiguo pasa a ser una columna normal o, con `drop=True`, se descarta. |

Un detalle práctico: al guardar con `.to_csv(...)`, pandas escribe también el índice como primera columna. Por eso en la Semana 1 guardarás con `index=False`, para que no se cuele la numeración de filas como si fuera un dato.

Míralo en los dos últimos casos de la tabla:

```python
print(altas)                          # filtrada: las filas conservan sus nombres originales (1 y 3)
print()
print(altas.reset_index(drop=True))   # renumeradas 0, 1; los nombres antiguos se descartan
print()
print(por_hora.reset_index())         # la hora sale del índice y vuelve a ser una columna
```

Y la forma **con bucle**, útil para entender qué hace `groupby` por dentro. Recuerda el desempaquetado de dos variables de la sección 4: en cada vuelta, `valor_hora` recibe el valor del grupo (`'14:00'`, `'17:00'`…) y `grupo` recibe sus filas. (En la práctica, y en la Semana 1, se usa la forma directa sin bucle que ves un poco más abajo —es más corta y más rápida—; el bucle es solo para entender el mecanismo.)

```python
for valor_hora, grupo in tabla.groupby('hora'):
    print("hora", valor_hora)
    print(grupo)
    print("sílice media de esa hora:", grupo['silice'].mean())
    print("-----")
```

Fíjate en lo que ocurre: el bucle da tres vueltas (14:00, 17:00 y 19:00); en cada una, `valor_hora` es un texto suelto con la hora (todavía no es un índice), `grupo` es la mini-tabla con solo las filas de esa hora, y `grupo['silice'].mean()` es la media de **esa** hora. Las filas de cada `grupo` conservan sus nombres de la tabla original (`0, 1`, luego `2, 3`, luego `4`): cada grupo es, en el fondo, un filtro de `tabla`. El `groupby` solo reparte las filas en grupos; el cálculo lo haces tú dentro del bucle. Es exactamente lo que hace `groupby` por dentro —pero en la Semana 1 no escribirás este bucle: usarás la forma directa de las celdas siguientes, que hace lo mismo para todas las columnas a la vez.

> **Mini-reto.** Con la `tabla` de arriba, (a) selecciona la columna `hora`, (b) filtra las filas con `silice` mayor o igual que 2.5, y (c) calcula la sílice media por `columna` (agrupa por `'columna'`).

**La forma que usarás en la Semana 1.** Si no eliges columna, `.mean()` calcula la media de **todas** las columnas en cada grupo, de una vez. Además, con `as_index=False` la hora **no** pasa al índice: se queda como una columna normal del resultado, y las filas se numeran 0, 1, 2… como en `tabla`. Es la misma información que arriba, colocada de otra forma, y así es más fácil juntar varios resúmenes en una tabla:

```python
medias = tabla.groupby('hora', as_index=False).mean()
medias
```

Dos resúmenes más con el mismo `groupby`: `.first()` se queda con el **primer** valor de cada grupo (útil cuando una columna es constante dentro del grupo) y `.size()` cuenta cuántas **filas** tiene cada grupo (en una columna llamada `size`):

```python
print(tabla.groupby('hora', as_index=False).first())
print()
print(tabla.groupby('hora', as_index=False).size())
```

Y para montar una tabla nueva con solo las columnas que te interesan, y con el nombre que quieras, se crea una tabla vacía y se llena columna a columna. Funciona porque todos los resúmenes salen del mismo `groupby`, con los grupos en el mismo orden: la fila 0 es la de las 14:00 en todos ellos. Es el patrón de la sección 4 de la Semana 1:

```python
filas = tabla.groupby('hora', as_index=False).size()

resumen = pd.DataFrame()
resumen['hora']         = medias['hora']
resumen['silice_media'] = medias['silice']
resumen['n_filas']      = filas['size']
resumen
```

**La forma compacta (opcional): `.agg()` con nombres.** Quien ya conozca pandas puede agrupar y calcular varias columnas-resumen a la vez, dando un nombre nuevo a cada una, en una sola instrucción. En la Semana 1 aparece solo en la ampliación:

```python
tabla.groupby('hora').agg(
    silice_media=('silice', 'mean'),
    silice_max=('silice', 'max'),
).reset_index()
```
